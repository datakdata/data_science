```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from matplotlib import rcParams
import warnings
warnings.filterwarnings('ignore')

# 设置中文字体
plt.rcParams['font.sans-serif'] = ['SimHei', 'Microsoft YaHei']
plt.rcParams['axes.unicode_minus'] = False

# 读取数据文件
print("正在读取数据文件...")

# 读取附件1
df_land = pd.read_excel('D:/pythonproject/data_science/excel_files/附件1.xlsx', sheet_name='乡村的现有耕地')
df_crops = pd.read_excel('D:/pythonproject/data_science/excel_files/附件1.xlsx', sheet_name='乡村种植的农作物')

# 读取附件2
df_planting_2023 = pd.read_excel('D:/pythonproject/data_science/excel_files/附件2.xlsx', sheet_name='2023年的农作物种植情况')
df_stats_2023 = pd.read_excel('D:/pythonproject/data_science/excel_files/附件2.xlsx', sheet_name='2023年统计的相关数据')

print("数据读取完成！")
print("\n附件1 - 耕地信息表:")
print(df_land.head())
print(f"耕地数据形状: {df_land.shape}")

print("\n附件1 - 农作物信息表:")
print(df_crops.head())
print(f"农作物数据形状: {df_crops.shape}")

print("\n附件2 - 2023年种植情况:")
print(df_planting_2023.head())
print(f"种植情况数据形状: {df_planting_2023.shape}")

print("\n附件2 - 2023年统计数据:")
print(df_stats_2023.head())
print(f"统计数据形状: {df_stats_2023.shape}")
```

```python
# 使用列索引进行数据预处理
print("使用列索引进行数据预处理...")

# 获取列名列表
land_columns = df_land.columns.tolist()
planting_columns = df_planting_2023.columns.tolist()
stats_columns = df_stats_2023.columns.tolist()

print(f"耕地数据列名: {land_columns}")
print(f"种植情况列名: {planting_columns}")
print(f"统计数据列名: {stats_columns}")

# 使用列索引访问数据
# 耕地面积在第3列（索引2）
# 种植面积在第5列（索引4）

# 1. 耕地数据分析
print("\n=== 耕地数据分析 ===")
print("耕地类型分布:")
print(df_land.iloc[:, 1].value_counts())  # 第2列是地块类型

# 计算各类耕地面积
total_land_area = df_land.iloc[:, 2].sum()  # 第3列是面积
print(f"\n总耕地面积: {total_land_area:.2f}亣")

land_type_area = df_land.groupby(df_land.iloc[:, 1])[df_land.columns[2]].sum()
print("\n各类耕地面积:")
print(land_type_area)

# 2. 农作物数据分析
print("\n=== 农作物数据分析 ===")
print("农作物类型分布:")
print(df_crops.iloc[:, 2].value_counts())  # 第3列是作物类型

# 3. 2023年种植情况分析
print("\n=== 2023年种植情况分析 ===")
print("2023年种植的农作物类型分布:")
print(df_planting_2023.iloc[:, 3].value_counts())  # 第4列是作物类型

print("\n2023年种植的农作物名称分布:")
print(df_planting_2023.iloc[:, 2].value_counts())  # 第3列是作物名称

# 4. 2023年统计数据分析
print("\n=== 2023年统计数据分析 ===")
print("统计数据中的农作物类型分布:")
print(df_stats_2023.iloc[:, 2].value_counts())  # 第3列是作物名称，需要查看作物类型在哪列

# 查看统计数据的列索引
print("\n统计数据列索引对应:")
for i, col in enumerate(stats_columns):
    print(f"索引 {i}: {col}")

# 处理销售单价数据（有些是范围）
def process_price(price_str):
    if isinstance(price_str, str) and '-' in price_str:
        prices = price_str.split('-')
        return (float(prices[0]) + float(prices[1])) / 2
    elif isinstance(price_str, str):
        return float(price_str)
    else:
        return price_str

df_stats_2023['平均销售单价'] = df_stats_2023.iloc[:, 7].apply(process_price)  # 第8列是销售单价

print("\n处理后的销售单价数据:")
print(df_stats_2023.iloc[:10, [2, 7, -1]])  # 显示作物名称、原始单价、平均单价
```
