```python
import pandas as pd
import numpy as np
import os
from itertools import product
import warnings
warnings.filterwarnings('ignore')

# 设置中文字体
import matplotlib.pyplot as plt
plt.rcParams['font.sans-serif'] = ['SimHei']  # 用来正常显示中文标签
plt.rcParams['axes.unicode_minus'] = False  # 用来正常显示负号

# 读取数据文件
file_path = r"D:\pythonproject\data_science\excel_files"

# 读取附件1
df_land = pd.read_excel(os.path.join(file_path, "附件1.xlsx"), sheet_name="乡村的现有耕地")
df_crops = pd.read_excel(os.path.join(file_path, "附件1.xlsx"), sheet_name="乡村种植的农作物")

# 读取附件2
df_planting_2023 = pd.read_excel(os.path.join(file_path, "附件2.xlsx"), sheet_name="2023年的农作物种植情况")
df_stats_2023 = pd.read_excel(os.path.join(file_path, "附件2.xlsx"), sheet_name="2023年统计的相关数据")

print("=== 数据预处理分析 ===")
print("\n1. 地块信息:")
print(f"地块总数: {len(df_land)}")
print(df_land.head())
print(f"\n地块类型分布:")
print(df_land['地块类型'].value_counts())
print(f"\n总耕地面积: {df_land['地块面积/亩'].sum():.1f}亩")

print("\n2. 作物信息:")
print(f"作物总数: {len(df_crops)}")
print(df_crops.head(10))
print(f"\n作物类型分布:")
print(df_crops['作物类型'].value_counts())

print("\n3. 2023年种植情况:")
print(f"种植记录总数: {len(df_planting_2023)}")
print(df_planting_2023.head())

print("\n4. 2023年统计数据:")
print(f"统计记录总数: {len(df_stats_2023)}")
print(df_stats_2023.head())

# 数据清洗和预处理
# 处理地块数据
df_land_clean = df_land.dropna(subset=['地块名称']).copy()

# 处理作物数据
df_crops_clean = df_crops.dropna(subset=['作物编号']).copy()

# 处理2023年种植数据
df_planting_2023_clean = df_planting_2023.dropna(subset=['地块名称']).copy()

# 处理统计数据
df_stats_2023_clean = df_stats_2023.dropna(subset=['作物编号']).copy()

# 提取销售价格范围并计算平均值
def parse_price_range(price_str):
    if pd.isna(price_str):
        return np.nan
    if isinstance(price_str, (int, float)):
        return float(price_str)
    if '-' in str(price_str):
        prices = str(price_str).split('-')
        try:
            return (float(prices[0]) + float(prices[1])) / 2
        except:
            return np.nan
    else:
        try:
            return float(price_str)
        except:
            return np.nan

df_stats_2023_clean['销售单价/元/斤'] = df_stats_2023_clean['销售单价/(元/斤)'].apply(parse_price_range)

print("\n=== 数据预处理完成 ===")
print(f"清洗后地块数: {len(df_land_clean)}")
print(f"清洗后作物数: {len(df_crops_clean)}")
print(f"清洗后种植记录数: {len(df_planting_2023_clean)}")
print(f"清洗后统计记录数: {len(df_stats_2023_clean)}")

# 保存预处理后的数据
df_land_clean.to_excel('preprocessed_land_data.xlsx', index=False)
df_crops_clean.to_excel('preprocessed_crops_data.xlsx', index=False)
df_planting_2023_clean.to_excel('preprocessed_planting_2023.xlsx', index=False)
df_stats_2023_clean.to_excel('preprocessed_stats_2023.xlsx', index=False)

print("\n预处理数据已保存到当前目录")
```

```python
# 重新运行数据预处理
import pandas as pd
import numpy as np
import os
import json

# 读取数据文件
file_path = r"D:\pythonproject\data_science\excel_files"

# 读取附件1
df_land = pd.read_excel(os.path.join(file_path, "附件1.xlsx"), sheet_name="乡村的现有耕地")
df_crops = pd.read_excel(os.path.join(file_path, "附件1.xlsx"), sheet_name="乡村种植的农作物")

# 读取附件2
df_planting_2023 = pd.read_excel(os.path.join(file_path, "附件2.xlsx"), sheet_name="2023年的农作物种植情况")
df_stats_2023 = pd.read_excel(os.path.join(file_path, "附件2.xlsx"), sheet_name="2023年统计的相关数据")

# 数据清洗和预处理
df_land_clean = df_land.dropna(subset=['地块名称']).copy()
df_crops_clean = df_crops.dropna(subset=['作物编号']).copy()
df_planting_2023_clean = df_planting_2023.dropna(subset=['地块名称']).copy()
df_stats_2023_clean = df_stats_2023.dropna(subset=['作物编号']).copy()

# 提取销售价格范围并计算平均值
def parse_price_range(price_str):
    if pd.isna(price_str):
        return np.nan
    if isinstance(price_str, (int, float)):
        return float(price_str)
    if '-' in str(price_str):
        prices = str(price_str).split('-')
        try:
            return (float(prices[0]) + float(prices[1])) / 2
        except:
            return np.nan
    else:
        try:
            return float(price_str)
        except:
            return np.nan

df_stats_2023_clean['销售单价/元/斤'] = df_stats_2023_clean['销售单价/(元/斤)'].apply(parse_price_range)

# 建立地块类型与作物类型的兼容性规则
land_type_compatibility = {
    '平旱地': ['粮食', '粮食（豆类）'],
    '梯田': ['粮食', '粮食（豆类）'],
    '山坡地': ['粮食', '粮食（豆类）'],
    '水浇地': ['粮食', '蔬菜', '蔬菜（豆类）'],
    '普通大棚': ['蔬菜', '蔬菜（豆类）', '食用菌'],
    '智慧大棚': ['蔬菜', '蔬菜（豆类）']
}

# 创建地块-作物兼容性矩阵
compatibility_matrix = {}
for land_type in land_type_compatibility:
    compatibility_matrix[land_type] = {}
    for crop_type in land_type_compatibility[land_type]:
        compatible_crops = df_crops_clean[df_crops_clean['作物类型'] == crop_type]['作物编号'].tolist()
        compatibility_matrix[land_type][crop_type] = compatible_crops

print("=== 地块-作物兼容性矩阵 ===")
for land_type, crop_types in compatibility_matrix.items():
    print(f"\n{land_type}:")
    for crop_type, crops in crop_types.items():
        print(f"  {crop_type}: {len(crops)}种作物")

# 计算各作物的预期销售量
def calculate_expected_sales(df_planting, df_stats):
    """根据2023年种植面积和产量计算预期销售量"""
    expected_sales = {}
    
    for crop_id in df_planting['作物编号'].unique():
        crop_name = df_planting[df_planting['作物编号'] == crop_id]['作物名称'].iloc[0]
        
        # 计算总种植面积
        total_area = df_planting[df_planting['作物编号'] == crop_id]['种植面积/亩'].sum()
        
        # 获取平均产量
        avg_yield = df_stats[df_stats['作物编号'] == crop_id]['亩产量/斤'].mean()
        
        # 计算预期销售量
        expected_sales[crop_id] = {
            'crop_name': crop_name,
            'expected_sales': total_area * avg_yield,
            'total_area_2023': total_area,
            'avg_yield': avg_yield
        }
    
    return expected_sales

expected_sales_2023 = calculate_expected_sales(df_planting_2023_clean, df_stats_2023_clean)

print("\n=== 2023年预期销售量 ===")
for crop_id, data in expected_sales_2023.items():
    print(f"{data['crop_name']} (编号{crop_id}): {data['expected_sales']:.0f}斤")

# 计算各作物的平均成本和价格
def calculate_cost_price(df_stats):
    """计算各作物的平均成本和价格"""
    cost_price_data = {}
    
    for crop_id in df_stats['作物编号'].unique():
        crop_data = df_stats[df_stats['作物编号'] == crop_id]
        crop_name = crop_data['作物名称'].iloc[0]
        
        avg_cost = crop_data['种植成本/(元/亩)'].mean()
        avg_price = crop_data['销售单价/元/斤'].mean()
        avg_yield = crop_data['亩产量/斤'].mean()
        
        cost_price_data[crop_id] = {
            'crop_name': crop_name,
            'avg_cost': avg_cost,
            'avg_price': avg_price,
            'avg_yield': avg_yield,
            'profit_per_mu': avg_yield * avg_price - avg_cost
        }
    
    return cost_price_data

cost_price_data = calculate_cost_price(df_stats_2023_clean)

print("\n=== 作物成本价格数据 ===")
for crop_id, data in cost_price_data.items():
    print(f"{data['crop_name']}: 成本{data['avg_cost']}元/亩, 价格{data['avg_price']:.2f}元/斤, 亩产{data['avg_yield']}斤, 每亩利润{data['profit_per_mu']:.0f}元")

# 转换键为字符串以避免JSON序列化问题
compatibility_matrix_str = {}
for land_type, crop_types in compatibility_matrix.items():
    compatibility_matrix_str[land_type] = {}
    for crop_type, crops in crop_types.items():
        compatibility_matrix_str[land_type][crop_type] = [int(crop) for crop in crops]

expected_sales_2023_str = {}
for crop_id, data in expected_sales_2023.items():
    expected_sales_2023_str[str(crop_id)] = {
        'crop_name': data['crop_name'],
        'expected_sales': float(data['expected_sales']),
        'total_area_2023': float(data['total_area_2023']),
        'avg_yield': float(data['avg_yield'])
    }

cost_price_data_str = {}
for crop_id, data in cost_price_data.items():
    cost_price_data_str[str(crop_id)] = {
        'crop_name': data['crop_name'],
        'avg_cost': float(data['avg_cost']),
        'avg_price': float(data['avg_price']),
        'avg_yield': float(data['avg_yield']),
        'profit_per_mu': float(data['profit_per_mu'])
    }

# 保存兼容性矩阵和基础数据
with open('compatibility_matrix.json', 'w', encoding='utf-8') as f:
    json.dump(compatibility_matrix_str, f, ensure_ascii=False, indent=2)

with open('expected_sales_2023.json', 'w', encoding='utf-8') as f:
    json.dump(expected_sales_2023_str, f, ensure_ascii=False, indent=2)

with open('cost_price_data.json', 'w', encoding='utf-8') as f:
    json.dump(cost_price_data_str, f, ensure_ascii=False, indent=2)

print("\n数据预处理完成！")
print("兼容性矩阵、预期销售量和成本价格数据已保存")
```
